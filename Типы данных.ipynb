_Задание 13: Простейшие алгоритмы группировки_

- Создайте список из 10 чисел.
- Сгруппируйте числа в этом списке по четности (четные и нечетные числа).
- Выведите две группы чисел: четные и нечетные.

Используйте цикл for и условия if num % 2 == 0

````python
import random
my_list=[random.randint(1,100) for i in range (10)]
print(my_list)
even_list=[]
odd_list=[]
for num in my_list:
  if num%2==0:
    even_list.append(num)
  else:
    odd_list.append(num)
print(even_list)  
print(odd_list)
# or
even_list = [num for num in my_list if num%2==0]
odd_list=[num for num in my_list if num%2!=0]
print(even_list)  
print(odd_list)
````

Напишите два варианта решения: с использованием функций max() и min(), а также без них

_Задание 15: Использование цикла while_
- Напишите программу, которая будет угадывать число. 
- Задайте случайное число от 1 до 100 (или любое другое фиксированное число). 
- Пусть пользователь вводит свои догадки, а программа должна подсказывать, больше или меньше введенное число по сравнению с загаданным. 
- Цикл while должен продолжаться до тех пор, пока пользователь не угадает число.

Используйте input() для получения догадки от пользователя. Не забудьте привести это значение к целому числу с помощью int()

````python
import random

def number_guessing_game():
    secret_number = random.randint(1, 100)
    print(secret_number)
    guess_count = 0
    print("Добро пожаловать в игру «Угадай число»!")
    print("Я загадал число от 1 до 100.")

    while True:
        try:
            guess = int(input("Ваша догадка: "))
            guess_count += 1
            if guess < secret_number:
                print("Слишком мало. Попробуйте еще раз.")
            elif guess > secret_number:
                print("Слишком много. Попробуйте еще раз.")
            else:
                print(f"Поздравляем! Вы угадали число {secret_number} за {guess_count} попыток.")
                break
        except ValueError:
            print("Недопустимый ввод. Введите целое число.")

number_guessing_game()    
````
