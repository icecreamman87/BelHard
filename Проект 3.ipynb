```python
import pandas as pd
import numpy as np
df=pd.read_csv(r"F:\Data science\Intensiv\октябрь 2025 ДЗ\Качество данных, домашнее задание.csv")
print(df.info())
missing_values=df.isnull().sum()
print(missing_values)
dublicated_values=df.duplicated().sum()
print(f'Количество дубликатов:{dublicated_values}')
unique_values_age = df['age'].unique()
print(unique_values_age)
print(df['age'].value_counts(dropna=False))
df['age'] = pd.to_numeric(df['age'], errors='coerce')
print(df['age'].value_counts(dropna=False))
age_abnormal = df[(df['age'] < 18) | (df['age'] > 100) | (df['age'].isnull())]
df['age'] = df['age'].mask((df['age'] < 18) | (df['age'] > 100))
print(f'Количество некорректных значений возраста:{age_abnormal.shape[0]}')
df['age'] = df['age'].fillna('unknown age')
print(df['age'].value_counts(dropna=False))
unique_values_email = df['email'].unique()
print(df['email'].value_counts(dropna=False))
df['email'] = df['email'].replace('invalid_email', np.nan)
df['email'] = df['email'].replace('charlie@example', 'charlie@example.com')
df['email'] = df['email'].replace('kevin', 'kevin@example.com')
print(df['email'].value_counts(dropna=False))
email_abnormal= df[df['email'].isnull()]
print(f'Количество некорректных значений email:{email_abnormal.shape[0]}')
df['email'] = df['email'].fillna('unknown@example.com')
print(df['email'].value_counts(dropna=False))
print(df['salary'].value_counts(dropna=False))
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
salary_abnormal = df[(df['salary'] <= 0) | (df['salary'].isnull())]
df['salary'] = df['salary'].mask(df['salary'] <= 0)
print(df['salary'].value_counts(dropna=False))
q1 = df['salary'].quantile(0.25)
q3 = df['salary'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
anomalous_values = df[(df['salary'] < lower_bound) | (df['salary'] > upper_bound)]
print(f'Количество некорректных значений salary:{salary_abnormal.shape[0]+anomalous_values.shape[0]}')
median_salary = df['salary'].median()
df['salary'] = np.where((df['salary'] < lower_bound) | (df['salary'] > upper_bound), median_salary, df['salary'])
df['salary'] = df['salary'].fillna('unknown salary')
print(df['salary'].value_counts(dropna=False))
print(df['join_date'].value_counts(dropna=False))
df['join_date'] = pd.to_datetime(df['join_date'],errors='coerce')
print(f'Количество некорректных значений date:{df['join_date'].isnull().sum()}')
df['join_date'] = df['join_date'].dt.strftime('%d-%m-%Y')
df['join_date'] = df['join_date'].fillna('unknown date')
print(df['join_date'].value_counts(dropna=False))
print(df.info())
new_file_path = r'F:\Data science\Intensiv\октябрь 2025 ДЗ\Качество данных, домашнее задание upd.csv'
df.to_csv(new_file_path)
```
