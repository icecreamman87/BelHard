```python
#Задание 4: Работа с пропущенными данными
import pandas as pd
df=pd.read_csv(r"F:\Data science\Intensiv\employees.csv")
print(df)
print(df.isnull().sum())
print(df.isnull().any(axis=1))
print(df.dropna())
print(df['Salary'].fillna(df['Salary'].mean()).reset_index())
```

```python
#Задание 5: Группировка данных
import pandas as pd
df=pd.read_csv(r"F:\Data science\Intensiv\data.csv")
df_group=df.groupby('City')['Salary'].mean().reset_index()
df_group.rename(columns={'Salary': 'Avr. salary'}, inplace=True)
print(df_group)
```

```python
#Задание 6: Объединение данных
import pandas as pd
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Anna', 'Ivan', 'Sergey']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Salary': [1000, 1500, 1200]
})
df3= df1.merge(df2)
print(df3)
print(pd.concat([df1, df2], axis=1))
print(pd.concat([df1, df2], axis=0))
```

```python
# Задание 7.1: Визуализация данных с Pandas
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv(r"F:\Data science\Intensiv\data.csv")
df['Salary'].plot(kind='hist', bins=30, edgecolor='red')
plt.title('Распределение зарплат')
plt.xlabel('Зарплата')
plt.ylabel('Частота')
```

```python
# Задание 7.2: Визуализация данных с Pandas
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv(r"F:\Data science\Intensiv\employees.csv")
df_empl=df['City'].value_counts()
print(df_empl)
df_empl.plot(kind='pie', autopct='%1.9f%%', startangle=90, colors=['skyblue', 'lightgreen', 'orange'])
plt.title('Распределение сотрудников')
plt.axis('equal')
```
