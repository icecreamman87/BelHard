```python
#Задание 13: Простейшие алгоритмы группировки
import random
my_list=[random.randint(1,100) for i in range (10)]
print(my_list)
even_list=[]
odd_list=[]
for num in my_list:
  if num%2==0:
    even_list.append(num)
  else:
    odd_list.append(num)
print(even_list)  
print(odd_list)
# or
even_list = [num for num in my_list if num%2==0]
odd_list=[num for num in my_list if num%2!=0]
print(even_list)  
print(odd_list)
```

```python
#Задание 14: Простейшие алгоритмы поиска
import random
my_list=[random.randint(-100,100) for i in range (20)]
print(my_list)
#способ 1
print(f" Мин.значение= {min(my_list)}, Макс.значение = {max(my_list)}")

# способ 2
my_list.sort()
print(f" Мин.значение ={my_list[0]}, Макс.значение = {my_list[-1]}")

# способ 3
max_1 = my_list[0]
min_1 = my_list[0]
for num in my_list:
    if num > max_1:
        max_1 = num
    if num < min_1:
        min_1 = num
print(f" Мин.значение ={min_1}, Макс.значение = {max_1}")
```

```python
#Задание 15: Использование цикла while
import random

def number_guessing_game():
    secret_number = random.randint(1, 100)
    print(secret_number)
    guess_count = 0
    print("Добро пожаловать в игру «Угадай число»!")
    print("Я загадал число от 1 до 100.")

    while True:
        try:
            guess = int(input("Ваша догадка: "))
            guess_count += 1
            if guess < secret_number:
                print("Слишком мало. Попробуйте еще раз.")
            elif guess > secret_number:
                print("Слишком много. Попробуйте еще раз.")
            else:
                print(f"Поздравляем! Вы угадали число {secret_number} за {guess_count} попыток.")
                break
        except ValueError:
            print("Недопустимый ввод. Введите целое число.")

number_guessing_game()            
```
